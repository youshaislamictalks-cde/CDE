Q22. Set Operations
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
Find union, intersection, a - b, and symmetric difference.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

union = a | b
intersection = a & b
difference = a - b
symmetric_difference = a ^ b

print("Union:", union)
print("Intersection:", intersection)
print("a - b:", difference)
print("Symmetric difference:", symmetric_difference)
```

Q25. Subset and Superset
required_skills = {"Python", "SQL"}

Check required skills are a subset and candidate skills a superset.

```python
required_skills = {"Python", "SQL"}
candidate_skills = {"Python", "SQL", "AWS", "Snowflake"}

is_subset = required_skills.issubset(candidate_skills)
is_superset = candidate_skills.issuperset(required_skills)

print("Required skills are a subset:", is_subset)
print("Candidate skills are a superset:", is_superset)
```

Q27. Three-Set Analysis
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}
Find students in all three, exactly Python and SQL (not AWS), and at least two.

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

all_three = python & sql & aws

exactly_python_sql = (python & sql) - aws

at_least_two = (python & sql) | (python & aws) | (sql & aws)

print("Students in all three:", all_three)
print("Exactly Python and SQL, not AWS:", exactly_python_sql)
print("Students in at least two:", at_least_two)
```

Q28. Transaction Analysis
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}
Find IDs appearing on all days, exactly one day, and at least two days.

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

all_days = day1 & day2 & day3

exactly_one_day = (
    (day1 - day2 - day3)
    | (day2 - day1 - day3)
    | (day3 - day1 - day2)
)

at_least_two_days = (
    (day1 & day2)
    | (day1 & day3)
    | (day2 & day3)
)

print("IDs appearing on all days:", all_days)
print("IDs appearing on exactly one day:", exactly_one_day)
print("IDs appearing on at least two days:", at_least_two_days)
```

Q29. Access Control
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}
Find missing and extra permissions and test full authorization.

```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

missing_permissions = required_permissions - user_permissions
extra_permissions = user_permissions - required_permissions
fully_authorized = required_permissions.issubset(user_permissions)

print("Missing permissions:", missing_permissions)
print("Extra permissions:", extra_permissions)
print("Full authorization:", fully_authorized)
```

Q30. Customer Segmentation
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}
Find customers on exactly one, exactly two, and all three platforms.

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website & mobile & store

exactly_two = (
    ((website & mobile) - store)
    | ((website & store) - mobile)
    | ((mobile & store) - website)
)

exactly_one = (
    (website - mobile - store)
    | (mobile - website - store)
    | (store - website - mobile)
)

print("Customers on exactly one platform:", exactly_one)
print("Customers on exactly two platforms:", exactly_two)
print("Customers on all three platforms:", all_three)
```
