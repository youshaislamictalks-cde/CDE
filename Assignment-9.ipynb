Q15. Extended Tuple Unpacking
data = (100, 200, 300, 400, 500, 600)
Unpack first two and last into variables; put all remaining elements into a list.

```python
data = (100, 200, 300, 400, 500, 600)

first, second, *remaining, last = data
remaining_list = list(remaining)

print("First:", first)
print("Second:", second)
print("Remaining elements:", remaining_list)
print("Last:", last)
```

Q17. Nested Tuple Access
students = (
 ("Ali", (85, 90, 88)),
 ("Sara", (92, 87, 95)),
 ("Ahmed", (78, 84, 80))
)
Extract Sara second score, Ali average, and Ahmed highest score.

```python
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

sara_second_score = students[1][1][1]
ali_average = sum(students[0][1]) / len(students[0][1])
ahmed_highest_score = max(students[2][1])

print("Sara second score:", sara_second_score)
print("Ali average:", ali_average)
print("Ahmed highest score:", ahmed_highest_score)
```

Q18. Tuple Sorting
employees = (
 ("Ali", 80000),
 ("Sara", 120000),
 ("Ahmed", 95000)
)
Sort by salary descending without loops, comprehensions, or lambda

```python
employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sorted_employees = tuple(
    sorted(employees, key=__import__("operator").itemgetter(1), reverse=True)
)

print("Employees sorted by salary descending:", sorted_employees)
```

Q19. Tuple Matrix
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))
Extract diagonal, calculate its product, and build a tuple of row sums.

```python
matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (
    matrix[0][0],
    matrix[1][1],
    matrix[2][2]
)

diagonal_product = (
    diagonal[0] * diagonal[1] * diagonal[2]
)

row_sums = (
    sum(matrix[0]),
    sum(matrix[1]),
    sum(matrix[2])
)

print("Diagonal:", diagonal)
print("Diagonal product:", diagonal_product)
print("Row sums:", row_sums)
```

Q20. Inventory Tuple Challenge
records = (
 ("Laptop", 120000, 5),
 ("Mouse", 1500, 20),
 ("Keyboard", 3500, 10)
)
Calculate inventory value, find most expensive product, and build reversed tuple of names.

```python
records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

inventory_value = (
    records[0][1] * records[0][2]
    + records[1][1] * records[1][2]
    + records[2][1] * records[2][2]
)

most_expensive = max(
    records,
    key=__import__("operator").itemgetter(1)
)

reversed_names = (
    records[2][0],
    records[1][0],
    records[0][0]
)

print("Total inventory value:", inventory_value)
print("Most expensive product:", most_expensive[0])
print("Reversed tuple of names:", reversed_names)
```
